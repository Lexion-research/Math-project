```python
import sympy as sp
import itertools
from sympy import symbols
```

Here we are trying to find exact constraints on coefficient functions such that a special classes of bivector fields on C^4 satisfies Schouten bracket condition.

```python
p = symbols('p_0, p_1, p_2, p_3')
t = symbols('t_0, t_1, t_2, t_3')

p0, p1, p2, p3 = p
t0, t1, t2, t3 = t

x = (t0, p0, t1, p1, t2, p2, t3, p3)
```

Each variable p_i and t_i plays the role of radial ($p_j=r_j^2$) and angular ( $\theta _j$ ) coordinates respectively in $\mathbb{C}^4$

```python
a=sp.Function('a')(*x)
b=sp.Function('b')(*x)
c=sp.Function('c')(*x)
e=sp.Function('e')(*x)
d=sp.Function('d')(*x)
g=sp.Function('g')(*x)
h=sp.Function('h')(*x)
r=sp.Function('r')(*x)
q=sp.Function('q')(*x)
s=sp.Function('s')(*x)
u=sp.Function('u')(*x)
v=sp.Function('v')(*x)
w=sp.Function('w')(*x)
```

Each of them is a smooth function on $\mathbb{C}^4$ that plays a role in defining the coefficient functions of the bivector field.

Below, we have used the smooth functions and variables to construct coefficient function and define bivector fields on $(\mathbb{C}^4)$ using skew symmetric matrices.

```python
N = sp.Matrix([
    # t0  p0  t1  p1  t2  p2  t3  p3
    [0,   0,  a,  0,  b,  0,  c,  0], # t0
    [0,   0,  0,  0,  0,  0,  0,  0], # p0
    [-a,  0,  0,  0,  w,  0,  v,  0], # t1
    [0,   0,  0,  0,  0,  0,  0,  0], # p1
    [-b,  0, -w,  0,  0,  0,  u,  0], # t2
    [0,   0,  0,  0,  0,  0,  0,  0], # p2
    [-c,  0, -v,  0, -u,  0,  0,  0], # t3
    [0,   0,  0,  0,  0,  0,  0,  0]  # p3
])
```

N describes the angular vector field $\Pi_1$.


```python
M = sp.Matrix([
    # t0   p0                  t1   p1                  t2   p2                  t3   p3
    [0,    0,                  0,    0,                  0,    0,                  0,    0],            # t0
    [0,    0,                  0,    -e * p0 * p1 ,      0,    g * p0 * p2,        0,    -h * p0 * p3], # p0
    [0,    0,                  0,    0,                  0,    0,                  0,    0],            # t1
    [0,    e * p0 * p1 ,       0,    0,                  0,    -d * p1 * p2,       0,    -r * p1 * p3], # p1
    [0,    0,                  0,    0,                  0,    0,                  0,    0],            # t2
    [0,    -g * p0  * p2,      0,    d * p1 * p2,        0,    0,                  0,    -q * p2 * p3], # p2
    [0,    0,                  0,    0,                  0,    0,                  0,    0],            # t3
    [0,    h * p0 * p3,        0,    r * p1 * p3,        0,    q * p2 * p3,        0,    0]             # p3
])

E = sp.Matrix([])
```

M describes the radial vector fields $\Pi_3$ and $\Pi^R_3$.

```python
# @title
P = sp.Matrix([
    # t0   p0                        t1   p1                          t2   p2                          t3   p3
    [0,    0,                        0,    0,                          0,    0,                          0,    0], # t0 (idx 0)
    [0,    0,                        0,    0 ,                     0,    g * p0 * p1 * p2 * p3,   0,    -h * p0 * p1 * p2 * p3], # p0 (idx 1)
    [0,    0,                        0,    0,                          0,    0,                          0,    0], # t1 (idx 2)
    [0,    0 ,                       0,    0,                          0,    -d * p0 * p1 * p2 * p3,  0,    -r  * p0 * p1 * p2 * p3], # p1 (idx 3)
    [0,    0,                        0,    0,                          0,    0,                          0,    0], # t2 (idx 4)
    [0,    -g * p0 * p1 * p2 * p3,  0,    d * p0 * p1 * p2 * p3,   0,    0,                          0,    0], # p2 (idx 5)
    [0,    0,                        0,    0,                          0,    0,                          0,    0], # t3 (idx 6)
    [0,    h * p0 * p1 * p2 * p3,   0,    r * p0 * p1 * p2 * p3,   0,    0,           0,    0]  # p3 (idx 7)
])

E = sp.Matrix([])
```

```python
# @title
N = sp.Matrix([
    # t0  p0  t1  p1  t2  p2  t3  p3
    [0,   0,  a,  0,  b,  0,  c,  0], # t0 (idx 0)
    [0,   0,  0,  0,  0,  0,  0,  0], # p0 (idx 1)
    [-a,  0,  0,  0,  w,  0,  v,  0], # t1 (idx 2)
    [0,   0,  0,  0,  0,  0,  0,  0], # p1 (idx 3)
    [-b,  0, -w,  0,  0,  0,  u,  0], # t2 (idx 4)
    [0,   0,  0,  0,  0,  0,  0,  0], # p2 (idx 5)
    [-c,  0, -v,  0, -u,  0,  0,  0], # t3 (idx 6)
    [0,   0,  0,  0,  0,  0,  0,  0]  # p3 (idx 7)
])
```

```python
P = M + N
display(P)
```

P describes the radial vector fields $\Pi_4$ and $\Pi^R_4$.

```python
def f(i, j, k):
    return sum(
        P[i,l]*P[j,k].diff(x[l])
        + P[j,l]*P[k,i].diff(x[l])
        + P[k,l]*P[i,j].diff(x[l]) for l in range(8))
```

Schouten bracket condition can be expressed in terms of Coefficient functions. The above expression computes the expressions that needs to be zero in order to satisfy Schouten brakcet condition.

The below code allows to add additional constraints on coeffecient functions and shows the final expression taking the additional constraints into account.

```python
for i, j, k in itertools.product(range(8), repeat=3):
    expr = f(i,j,k).simplify().subs({
        # p0 * a.diff(p0): -a,
        # p1 * a.diff(p1): -a,
        # a.diff(p2): 0,

        # b.diff(p0): 0,
        # p1 * b.diff(p1): -b,
        # p2 * b.diff(p2): -b,

        # p0 * c.diff(p0): -c,
        # c.diff(p1): 0,
        # p2 * c.diff(p2): -c,
         a.diff(t0): 0,
         a.diff(t1): 0,
         a.diff(t2): 0,
         a.diff(t3): 0,
        a.diff(p0): 0,
        a.diff(p1): 0,
        a.diff(p2): 0,
         a.diff(p3): 0,
        b.diff(p0): 0,
        b.diff(p1): 0,
        b.diff(p2): 0,
         b.diff(p3): 0,
        c.diff(p0): 0,
        c.diff(p1): 0,
        c.diff(p2): 0,
         c.diff(p3): 0,

         b.diff(t0): 0,
         b.diff(t1): 0,
         b.diff(t2): 0,
         b.diff(t3): 0,

         c.diff(t0): 0,
         c.diff(t1): 0,
         c.diff(t2): 0,
         c.diff(t3): 0,

         e.diff(t0): 0,
          e.diff(t1): 0,
         e.diff(p0): 0,
         e.diff(p1): 0,
       e.diff(p2): 0,
          e.diff(t2): 0,
         e.diff(t3): 0,
         e.diff(p3): 0,


          d.diff(t1): 0,
          d.diff(t2): 0,
         d.diff(p0): 0,
         d.diff(p1): 0,
         d.diff(p2): 0,
          d.diff(t0): 0,
         d.diff(t3): 0,
         d.diff(p3): 0,


          g.diff(t0): 0,
          g.diff(t2): 0,
         g.diff(p1): 0,
         g.diff(p2): 0,
         g.diff(p0): 0,
          g.diff(t1): 0,
         g.diff(t3): 0,
         g.diff(p3): 0,

         h.diff(t0): 0,
         h.diff(t1): 0,
         h.diff(t2): 0,
         h.diff(t3): 0,
         h.diff(p0): 0,
         h.diff(p1): 0,
         h.diff(p2): 0,
         h.diff(p3): 0,

         q.diff(t0): 0,
         q.diff(t1): 0,
         q.diff(t2): 0,
         q.diff(t3): 0,
         q.diff(p0): 0,
         q.diff(p1): 0,
         q.diff(p2): 0,
         q.diff(p3): 0,

         r.diff(t0): 0,
         r.diff(t1): 0,
         r.diff(t2): 0,
         r.diff(t3): 0,
         r.diff(p0): 0,
         r.diff(p1): 0,
         r.diff(p2): 0,
         r.diff(p3): 0,

         s.diff(t0): 0,
         s.diff(t1): 0,
         s.diff(t2): 0,
         s.diff(t3): 0,
         s.diff(p0): 0,
         s.diff(p1): 0,
         s.diff(p2): 0,
         s.diff(p3): 0,

         u.diff(t0): 0,
         u.diff(t1): 0,
         u.diff(t2): 0,
         u.diff(t3): 0,
         u.diff(p0): 0,
         u.diff(p1): 0,
         u.diff(p2): 0,
         u.diff(p3): 0,

         v.diff(t0): 0,
         v.diff(t1): 0,
         v.diff(t2): 0,
         v.diff(t3): 0,
         v.diff(p0): 0,
         v.diff(p1): 0,
         v.diff(p2): 0,
         v.diff(p3): 0,

         w.diff(t0): 0,
         w.diff(t1): 0,
         w.diff(t2): 0,
         w.diff(t3): 0,
         w.diff(p0): 0,
         w.diff(p1): 0,
         w.diff(p2): 0,
         w.diff(p3): 0
    })
    if expr:
        print(i, j, k)
        display(expr)
```
